```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load dataset
df = pd.read_csv('dataset/kdramas.csv')
```

```python
# drop column 'image_url' since it is completely not necessary for EDA
df.drop('image_url', axis=1, inplace=True)
df.head(10)
```

```python
print(f"There are {df.shape[0]} Korean dramas in the dataset.")
print(f"There are {df['title'].nunique()} unique Korean drama titles in the dataset.")
```

```python
print(f"There are {df.duplicated().sum()} Korean dramas in the dataset.")
```

#### *It is really important to highlight here : Although there are some dramas having similar titles, there are no duplicated dramas in the dataset. Therefore, it is not necessary to drop any rows/dramas in the dataset based on its title.*

```python
df.isnull().sum()
```

```python
# For categorical columns, I will fill it with a placeholder string
df['age_rating'] = df['age_rating'].fillna('Unknown')
df['mainLead1'] = df['mainLead1'].fillna('Unknown')
df['mainLead2'] = df['mainLead2'].fillna('Unknown')
df['synopsis'] = df['synopsis'].fillna('Unknown')
```

```python
# For any drama with null values for endYear, I will replace it with its startYear and if their startYear is missing, I fill it with endYear
print("Missing endYears: ", (df['endYear'] == '\\N').sum())
print("Missing startYears: ", (df['startYear'] == '\\N').sum())
```
